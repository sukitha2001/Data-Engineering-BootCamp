```python
# Lmbda functions

def add(x, y):
    return x + y

add(2,3)
```

```python
addition = lambda x, y: x + y
print(addition(2, 3))
```

```python
def even(num):
    if num % 2 == 0:
        return True

is_even = lambda num: num % 2 == 0

print(is_even(4))
print(even(4))
```

```python
# Map Functions

def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5]
squared_numbers = map(square, numbers)
print(list(squared_numbers))
```

```python
# filter function

def is_even(num):
    return num % 2 == 0

even_func = lambda num: num % 2 == 0
```

```python
lst1 = [1, 2, 3, 4, 5]
even_numbers = filter(is_even, lst1)
print(list(even_numbers))
```

```python
even_numbers = filter(even_func, lst1)
print(list(even_numbers))
```
